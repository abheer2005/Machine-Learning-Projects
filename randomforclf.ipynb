```python
import pandas as pd
from sklearn.datasets import load_digits
digits=load_digits()
```

```python
dir(digits)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.gray()
for i in range(4):
  plt.matshow(digits.images[i])
```

```python
digits.data[:5]
```

```python
df=pd.DataFrame(digits.data)
df.head()
```

```python
digits.target
```

```python
df['target']=digits.target       #created new col in dataframe
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(df.drop(['target'],axis='columns'),digits.target,test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
model=RandomForestClassifier()
model.fit(X_train,Y_train)
```

```python
model.score(X_test,Y_test)
```

```python
y_predicted=model.predict(X_test)
```

**CONFUSION MATRIX TO SEE WHERE MODEL FAILED DID NOT DO WELL**

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(Y_test,y_predicted)
cm
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sn
plt.figure(figsize=(10,7))
sn.heatmap(cm,annot=True)
plt.xlabel("predicted")
plt.ylabel("truth")
```

**Exercise of iris dataset using randomforestclf**

```python
from sklearn.datasets import load_iris
iris=load_iris()
```

```python
dir(iris)
```

```python
iris.data[:5]
```

```python
d=pd.DataFrame(iris.data)
d.head()
```

```python
iris.target
```

```python
iris.target_names
```

```python
d['target']=iris.target       #created new col in dataframe
d.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(d.drop(['target'],axis='columns'),iris.target,test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
model=RandomForestClassifier(n_estimators=40)
model.fit(X_train,Y_train)
```

```python
model.score(X_test,Y_test)
```

```python
y_predicted=model.predict(X_test)
```

**CONFUSION MATRIX**

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(Y_test,y_predicted)
cm
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sn
plt.figure(figsize=(10,7))
sn.heatmap(cm,annot=True)
plt.xlabel("predicted")
plt.ylabel("truth")
```
