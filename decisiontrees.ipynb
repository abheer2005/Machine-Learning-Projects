```python
import pandas as pd
df=pd.read_csv("salariesdecisiontrees.csv")
df.head()
```

```python
inputs=df.drop('salary_more_then_100k',axis='columns')          #drop target column target col is whether salary is more than 100k or not
target=df['salary_more_then_100k']
```

```python
inputs
```

```python
target
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le_company=LabelEncoder()
le_job=LabelEncoder()
le_degree=LabelEncoder()
```

```python
inputs['company_n']=le_company.fit_transform(inputs['company'])
inputs['job_n']=le_job.fit_transform(inputs['job'])
inputs['degree_n']=le_degree.fit_transform(inputs['degree'])
inputs.head()
```

```python
inputs_n=inputs.drop(['company','job','degree'],axis='columns')
inputs_n
```

```python
from sklearn import tree
```

```python
model=tree.DecisionTreeClassifier()
```

```python
model.fit(inputs_n,target)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(inputs_n,target,test_size=0.2)
```

```python
model.score(X_test,Y_test)
```

**Exercise to find survival of titanic passengers by dec tree**

```python
d=pd.read_csv("titanic.csv")
d.head()
```

```python
d.drop(['PassengerId','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis='columns',inplace=True)
```

```python
d.head()
```

```python
inputs = d.drop('Survived',axis='columns')
target = d.Survived
```

```python
inputs.Sex = inputs.Sex.map({'male': 1, 'female': 2})
inputs.Age[:10]
```

```python
inputs.Age = inputs.Age.fillna(inputs.Age.mean())
inputs.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(inputs,target,test_size=0.2,random_state=20)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn import tree
model = tree.DecisionTreeClassifier()
```

```python
model.fit(X_train,Y_train)
```

```python
model.score(X_test,Y_test)
```
